```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python

import os
from google.colab import files

file_path = "/content/Unemployment in India.csv"

if not os.path.exists(file_path):
    uploaded = files.upload()
    file_path = next(iter(uploaded))

df = pd.read_csv(file_path)

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python

df_clean = df.copy()

# Convert date to datetime format
df_clean["Date"] = pd.to_datetime(
    df_clean["Date"],
    dayfirst=True,
    errors="coerce"
)

# Remove rows with missing or invalid values
df_clean = df_clean.dropna().copy()

# Sort records by date
df_clean = df_clean.sort_values("Date").reset_index(drop=True)

print("Data cleaning completed!")
print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

display(df_clean.head())
```

```python

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df_clean,
    x="Estimated Unemployment Rate (%)",
    bins=25,
    kde=True
)

plt.title("Distribution of Unemployment Rate in India")
plt.xlabel("Estimated Unemployment Rate (%)")
plt.ylabel("Number of Observations")

plt.tight_layout()
plt.savefig("unemployment_distribution.png", dpi=300)
plt.show()
```

```python

area_avg = (
    df_clean.groupby("Area")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Unemployment Rate by Area:")
display(area_avg.round(2))

plt.figure(figsize=(7, 5))

sns.barplot(
    data=df_clean,
    x="Area",
    y="Estimated Unemployment Rate (%)",
    errorbar=None
)

plt.title("Average Unemployment Rate: Rural vs Urban")
plt.xlabel("Area")
plt.ylabel("Average Unemployment Rate (%)")

plt.tight_layout()
plt.savefig("area_comparison.png", dpi=300)
plt.show()
```

```python

monthly_data = (
    df_clean.set_index("Date")
    ["Estimated Unemployment Rate (%)"]
    .resample("MS")
    .mean()
    .dropna()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_data.index,
    monthly_data.values,
    marker="o"
)

plt.title("Monthly Unemployment Rate Trend in India")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("monthly_trend.png", dpi=300)
plt.show()
```

```python

region_avg = (
    df_clean.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Unemployment Rate by Region:")
display(region_avg.round(2))

plt.figure(figsize=(10, 8))

region_avg.sort_values().plot(kind="barh")

plt.title("Average Unemployment Rate by Region")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.tight_layout()
plt.savefig("region_comparison.png", dpi=300)
plt.show()
```

```python

df_clean["Period"] = np.where(
    df_clean["Date"] < pd.Timestamp("2020-03-01"),
    "Before COVID-19",
    "March 2020 Onward"
)

covid_summary = (
    df_clean.groupby("Period")["Estimated Unemployment Rate (%)"]
    .agg(["mean", "median", "count"])
    .round(2)
)

print("Unemployment Comparison:")
display(covid_summary)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_clean,
    x="Period",
    y="Estimated Unemployment Rate (%)"
)

plt.title("Unemployment Rates Before and During COVID-19 Period")
plt.xlabel("")
plt.ylabel("Estimated Unemployment Rate (%)")

plt.tight_layout()
plt.savefig("covid_comparison.png", dpi=300)
plt.show()
```

```python

df_clean["Month"] = df_clean["Date"].dt.month

month_avg = (
    df_clean.groupby("Month")["Estimated Unemployment Rate (%)"]
    .mean()
    .reindex(range(1, 13))
)

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

month_avg.index = month_names

print("Average Unemployment Rate by Month:")
display(month_avg.round(2))

plt.figure(figsize=(11, 5))

month_avg.plot(kind="bar")

plt.title("Average Unemployment Rate by Month")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("monthly_seasonal_patterns.png", dpi=300)
plt.show()
```

```python

print("=" * 45)
print("UNEMPLOYMENT ANALYSIS - KEY FINDINGS")
print("=" * 45)

print("\nTotal records after cleaning:", len(df_clean))

print(
    "\nOverall average unemployment rate:",
    round(df_clean["Estimated Unemployment Rate (%)"].mean(), 2),
    "%"
)

print(
    "\nHighest average regional rate:",
    region_avg.idxmax(),
    "-",
    round(region_avg.max(), 2),
    "%"
)

print(
    "\nLowest average regional rate:",
    region_avg.idxmin(),
    "-",
    round(region_avg.min(), 2),
    "%"
)

print("\nAverage rate by area:")
print(area_avg.round(2))

print("\nCOVID-period comparison:")
display(covid_summary)

print("\nAnalysis completed successfully!")
```

```python

import zipfile
from google.colab import files

graph_files = [
    "unemployment_distribution.png",
    "area_comparison.png",
    "monthly_trend.png",
    "region_comparison.png",
    "covid_comparison.png",
    "monthly_seasonal_patterns.png"
]

with zipfile.ZipFile("unemployment_graphs.zip", "w") as zip_file:
    for graph in graph_files:
        if os.path.exists(graph):
            zip_file.write(graph)

files.download("unemployment_graphs.zip")
```
